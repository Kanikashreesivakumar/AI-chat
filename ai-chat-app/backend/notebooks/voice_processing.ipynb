```python
# Voice Processing Notebook

import requests
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile

# Function to record audio
def record_audio(duration=5, filename='output.wav'):
    import sounddevice as sd
    print('Recording...')
    audio = sd.rec(int(duration * 44100), samplerate=44100, channels=2)
    sd.wait()  # Wait until recording is finished
    wavfile.write(filename, 44100, audio)
    print('Recording complete.')

# Function to send audio to Whisper API
def transcribe_audio(filename):
    url = 'https://api.whisper.ai/transcribe'
    with open(filename, 'rb') as f:
        response = requests.post(url, files={'file': f})
    return response.json()

# Function to visualize audio waveform
def plot_waveform(filename):
    samplerate, data = wavfile.read(filename)
    plt.figure(figsize=(10, 4))
    plt.plot(data)
    plt.title('Audio Waveform')
    plt.xlabel('Samples')
    plt.ylabel('Amplitude')
    plt.show()

# Example usage
if __name__ == '__main__':
    record_audio()
    transcription = transcribe_audio('output.wav')
    print('Transcription:', transcription)
    plot_waveform('output.wav')
```
