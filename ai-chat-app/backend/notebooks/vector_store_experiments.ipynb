# Vector Store Experiments

This notebook is designed for experimenting with the vector store functionality using Redis/ChromaDB. 
We will explore how to store and retrieve vector embeddings for chat messages and user interactions.

```python
# Import necessary libraries
import redis
from langchain.embeddings import OpenAIEmbeddings
from langchain.vectorstores import Chroma

# Initialize Redis connection
redis_client = redis.StrictRedis(host='localhost', port=6379, db=0)

# Initialize embeddings model
embeddings_model = OpenAIEmbeddings()

# Function to store vector in Redis
def store_vector(key, vector):
    redis_client.set(key, vector)

# Function to retrieve vector from Redis
def retrieve_vector(key):
    return redis_client.get(key)

# Example usage
example_key = 'chat_message_1'
example_vector = embeddings_model.embed('Hello, how can I help you?')
store_vector(example_key, example_vector)
retrieved_vector = retrieve_vector(example_key)
print(f'Retrieved vector for {example_key}: {retrieved_vector}')
```
